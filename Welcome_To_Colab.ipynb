Business Scenario: ByteCorner Computer Accessories Shop

The business sells computer accessories such as gaming mice, keyboards, headsets, SSDs, RAM, and gaming monitors.

```
#Step 1. Import the libraries
#Create a code cell and copy:

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```
#Step 2. Create the business variables
products = ["Gaming Mouse", "Keyboard", "Headset",
            "SSD 1TB", "RAM 16GB", "Gaming Monitor"]

prices = {
    "Gaming Mouse": 850,
    "Keyboard": 1850,
    "Headset": 1250,
    "SSD 1TB": 3200,
    "RAM 16GB": 2400,
    "Gaming Monitor": 7800
}

print("Business: ByteCorner Computer Accessories Shop")
print("Products:", products)
print("Prices:", prices)
```

```
#Step 3. Generate 500 transactions
np.random.seed(42)

number_of_transactions = 500

selected_products = np.random.choice(products, 500)
quantities = np.random.choice([1, 2, 3], 500)
discounts = np.random.choice([0, 0.05, 0.10], 500)

dataset = pd.DataFrame({
    "Transaction_ID": range(1, 501),
    "Product": selected_products,
    "Unit_Price": [prices[p] for p in selected_products],
    "Quantity": quantities,
    "Discount": discounts
})

dataset["Transaction_Value"] = (
    dataset["Unit_Price"] *
    dataset["Quantity"] *
    (1 - dataset["Discount"])
)

dataset.head(10)
```

```
#Step 4. Display the dataset
print("Total Transactions:", len(dataset))

display(dataset)
```

```
#Step 5. Calculate total revenue
total_revenue = dataset["Transaction_Value"].sum()

print("Total Revenue: PHP", round(total_revenue, 2))
```

```
#Step 6. Calculate mean, median, and mode
values = dataset["Transaction_Value"]

print("Mean:", values.mean())
print("Median:", values.median())
print("Mode:", values.mode().tolist())
```

```
#Step 7. Calculate standard deviation
print("Population Standard Deviation:", values.std(ddof=0))

print("Sample Standard Deviation:", values.std(ddof=1))
```

```
#Step 8. Find the product with the highest revenue
product_revenue = dataset.groupby("Product")["Transaction_Value"].sum()

print(product_revenue)

print("\nHighest Revenue Product:")
print(product_revenue.idxmax())
```

```
#Step 9. Create a bar chart
product_revenue.sort_values().plot(
    kind="bar",
    figsize=(9,5)
)

plt.title("Total Revenue by Product")
plt.xlabel("Products")
plt.ylabel("Revenue (PHP)")
plt.show()
```

```
#Step 10. Create a histogram
plt.figure(figsize=(9,5))

plt.hist(values, bins=15, edgecolor="black")

plt.title("Distribution of Transaction Values")
plt.xlabel("Transaction Value (PHP)")
plt.ylabel("Number of Transactions")

plt.show()
```

Business Insights
1. The shop can identify which products generate the highest revenue and prioritize their stocks.
2. The mean, median, and mode help the owner understand how much customers usually spend.
3. The bar chart helps compare the total revenue of different products.
4. The histogram shows the distribution of customer purchases and helps the owner understand sales patterns.
